```python
#Day-1
import numpy as np

#Vectors
a = np.array([2,3])
b = np.array([1,-1])

#Vector Addition
print(a+b)

#Scalar Multiplication
print(3*a)

#Visualize vectors using Matplotlib
import matplotlib.pyplot as plt

origin = [0,0]
plt.quiver(*origin, *a, scale=1,scale_units='xy', angles='xy', color='r', label='a')
plt.quiver(*origin, *b, scale=1, scale_units='xy', angles='xy', color='b', label='b')
plt.quiver(*origin, *(a+b), scale=1, scale_units='xy', angles='xy',color='g', label='a+b')
plt.xlim(-1,5)
plt.ylim(-2,5)
plt.grid()
plt.legend()
plt.show()
```

```python
#Day-2
import numpy as np

#Define two matrices
a = np.array([[1,2],[3,4]])

b = np.array([[5,6],[7,8]])

#Matrix multiplication
print(1,a@b) #Preferred way in research code
print(2,np.dot(a,b)) #Same result

#Matrix times vector
v = np.array([1,2])
print(3,a@v)

#Notice - order matters
print(4,a@b)
print(5,b@a)

c = np.eye(2)
print(c@a)
```

```python
#Day-3
#Exercise-1
#Matrix multiplication without using np.dot() or @:
import numpy as np
A = np.array([[1,2],[3,4]])
B = np.array([[5,6],[7,8]])

#Manual matrix multiplication using loops
result = np.zeros((2,2))
for i in range(2):
    for j in range(2):
        for k in range(2):
            result[i][j] += A[i][k] * B[k][j]

print("Manual: ", result)
print("Numpy: ", A@B) #Verify the match
```

```python
#Exercise-2
def manual_transpose(M):
    rows,cols = M.shape
    result = np.zeros((cols,rows))
    for i in range(rows):
        for j in range(cols):
            result[j][i] = M[i][j]
    return result

print(manual_transpose(A))
print(A.T) #Verify
```

```python
#Exercise-3
def manual_dot(a,b):
    result = 0
    for i in range(len(a)):
        result += a[i] * b[i]
    return result

a = np.array([1,2,3])
b = np.array([4,5,6])
print(manual_dot(a,b))
print(np.dot(a,b)) #Verify
```

```python
#Exercise-4
import numpy as np
A = np.array(([1,2,3],
              [4,5,6]))
v = np.array([1,2,3])

result = np.zeros(2)
for i in range(A.shape[0]):
    for j in range(A.shape[1]):
        result[i] += A[i][j] * v[j]

print("Manual: ", result)
print("Numpy: ", A@v)
```

```python
#Day-5
import numpy as np

A = np.array([[3,1],
              [0,2]])

#Get eigenvalues and eigenvectors
eigenvalues,eigenvectors = np.linalg.eig(A)

print("Eigenvalues:",eigenvalues)
print("Eigenvectors:\n",eigenvectors)

#Verify - A@v should equal lambda * v
v = eigenvectors[:,0] #First eigenvector
lam = eigenvalues[0] #First eigenvalue

print("A@v:",A @ v)
print("lambda*v",lam*v) #should match
```

```python
import matplotlib.pyplot as plt
import numpy as np

A = np.array([[3,1],
              [0,2]])
eigenvalues,eigenvectors = np.linalg.eig(A)

origin=[0,0]
v = eigenvectors[:,0] #First eigenvector


#Original eigenvector
plt.quiver(*origin, *v, scale=1,scale_units='xy',angles='xy',color='b',label='eigenvector')

#Transformed eigenvector
plt.quiver(*origin, *(A@v), scale=1,scale_units='xy',angles='xy',color='r',label='A@v')

plt.xlim(-5,5)
plt.ylim(-5,5)
plt.grid()
plt.legend()
plt.title("Eigenvector before and after transformation")
plt.show()
```

```python
#Day-6
#Exercise-1
import numpy as np
import matplotlib.pyplot as plt

#Flip a coin 1000 times (0=tails, 1=heads)
flips = np.random.randint(0,2,1000)
print("Heads:",np.sum(flips))
print("Tails:",1000-np.sum(flips))
print("P(heads):",np.mean(flips)) #Should be close to 0.5
```

```python
#Exercise-2
import numpy as np
import matplotlib.pyplot as plt

#Rolling a dice 1000 times
rolls = np.random.randint(1,7,10000)

#Plot distribution
plt.hist(rolls,bins=6,edgecolor='black')
plt.title('Dice Roll Distribution')
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.show()

#each value should appear ~1667 times
for i in range(1,7):
    print(f"P({i}) = {np.mean(rolls == i):.3f}")
```

```python
#Exercise-3
#Generating and plotting normal distribution
import numpy as np
import matplotlib.pyplot as plt
data = np.random.normal(loc=0,scale=1,size=10000)

plt.hist(data,bins=50,edgecolor='black')
plt.title('Normal distribution (mean=0,std=1)')
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.show()

print("Mean:",np.mean(data)) #should be close to 0
print("Std:",np.std(data)) #should be close to 1S
```
