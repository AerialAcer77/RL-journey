```python
#defaultfmt argument
#If names=None but a structured dtype is expected, names are defined with the standard NumPy default of "f%i", yielding names like f0, f1 and so forth:
data = StringIO("1 2 3\n 4 5 6")
print(np.genfromtxt(data,dtype=(np.int_,np.float64,np.int_)))

#In the same way, if we don’t give enough names to match the length of the dtype, the missing names will be defined with this default template
data = StringIO("1 2 3\n 4 5 6")
print(np.genfromtxt(data, dtype=(np.int_, np.float64, np.int_), names="a"))

#We can overwrite this default with the defaultfmt argument, that takes any format string
data = StringIO("1 2 3\n 4 5 6")
np.genfromtxt(data,dtype=(np.int_,np.float64,np.int_),defaultfmt="var%02i")
```

```python
import numpy as np
from io import StringIO
#Tweaking the conversion
#converters argument
convertfunc = lambda x: float(x.strip("%"))/100
data = "1, 2.3%, 45.\n6, 78.9%, 0"
names = ("i", "p", "n")
#General case
print(np.genfromtxt(StringIO(data), delimiter=",",names=names))
''' We need to keep in mind that by default, dtype=np.float64. 
A float is therefore expected for the second column. 
However, the strings ' 2.3%' and ' 78.9%' cannot be converted to float and we end up having np.nan instead.'''

#Converted case
print(np.genfromtxt(StringIO(data),delimiter=",",names=names,converters={1:convertfunc}))

#Using a name for the converter:
print(np.genfromtxt(StringIO(data),delimiter=",",names=names,converters={"p":convertfunc}))

#filling spaces
data = "1, , 3\n 4, 5, 6"
convertfunc = lambda x: float(x.strip() or -999)
np.genfromtxt(StringIO(data),delimiter=",",converters={1:convertfunc})
```

```python
import numpy as np
#Broadcasting

#NumPy operations are usually done on pairs of arrays on an element-by-element basis. In the simplest case, the two arrays must have exactly the same shape
a = np.array([1.0,2.0,3.0])
b = np.array([2.0,2.0,2.0])
print(a*b)

#The simplest broadcasting example occurs when an array and a scalar value are combined in an operation
c = 2.0
a*c
```

```python
#General Broadcasting Rules
''' When operating on two arrays, NumPy compares their shapes element-wise. It starts with the trailing (i.e. rightmost) dimension and works its way left. Two dimensions are compatible when

1) they are equal, or

2) one of them is 1.

If these conditions are not met, a ValueError: operands could not be broadcast together exception is thrown, indicating that the arrays have incompatible shapes.'''

#An example of broadcasting when a 1-d array is added to a 2-d array:
a = np.array([[ 0.0,  0.0,  0.0],
              [10.0, 10.0, 10.0],
              [20.0, 20.0, 20.0],
              [30.0, 30.0, 30.0]])
b = np.array([1.0,2.0,3.0])
print(a+b)

b = np.array([1.0,2.0,3.0,4.0])
#print(a+b) ValueError

#The following example shows an outer addition operation of two 1-d arrays:
a = np.array([0.0,10.0,20.0,30.0])
b  = np.array([1.0,2.0,3.0])
print(a[:,np.newaxis] + b)
```

```python
'''Copies and Views
View - The data buffer remains the same, so any changes made to a view reflects in the original copy.
Copy - When a new array is created by duplicating the data buffer as well as the metadata, it is called a copy. 
       Changes made to the copy do not reflect on the original array. Making a copy is slower and memory-consuming but sometimes necessary.
'''

#Views are created when elements can be addressed with offsets and strides in the original array. Hence, basic indexing always creates views.
x = np.arange(10)
print(x)

y = x[1:3]
print(y)

x[1:3] = [10,11]
print(x)

y
#Here, y gets changed when x is changed because it is a view.
```

```python
x = np.arange(9).reshape(3,3)
print(x)
y = x[[1,2]]
print(y)

y.base is None
#Here, y is a copy, as signified by the base attribute. We can also confirm this by assigning new values to x[[1, 2]] which in turn will not affect y at all

x[[1,2]] = [[10,11,12],[13,14,15]]
print(x)
y
#It must be noted here that during the assignment of x[[1, 2]] no view or copy is created as the assignment happens in-place.
```

```python
#how to tell if the array is a view or a copy
#The base attribute of the ndarray makes it easy to tell if an array is a view or a copy. The base attribute of a view returns the original array while it returns None for a copy.
x = np.arange(9)
print(x)

y = x.reshape(3,3)
print(y)
print(y.base)

z = y[[2,1]]
print(z)
z.base is None
```

```python
#Code segment showing the “can cast safely” table for a 64-bit system. Generally the output depends on the system;
mark = {False:' -',True:' Y'}
def print_table(ntypes):
    print('X' + ' '.join(ntypes))
    for row in ntypes:
        print(row,end='')
        for col in ntypes:
            print(mark[np.can_cast(row,col)],end='')
        print()

print_table(np.typecodes['All'])
```

```python
#Program-3
#Create two arrays x = np.linspace(0, 10, 100) and plot x², x³ and √x on the same graph with a legend
import numpy as np
import matplotlib.pyplot as plt
x = np.linspace(0,10,100)
plt.plot(x, x**2, 'r^', label='x²')
plt.plot(x, x**3, 'bo', label='x³')
plt.plot(x, x**0.5, 'g-', label='√x')
plt.legend('tcl')
plt.show()
```

```python
#Program-5
#Generate two 3x3 matrices, add them, subtract them and plot both results side by side using plt.subplot()
import numpy as np
import matplotlib.pyplot as plt
c = np.arange(9).reshape(3,3)
d = np.linspace(1,50,9).reshape(3,3)

e = c+d
f = c-d
plt.subplot(1,2,1)
plt.imshow(e)
plt.subplot(1,2,2)
plt.imshow(f)
plt.tight_layout()
plt.colorbar()
plt.plot()
```

```python
#Program-6
#Create a scatter plot where color represents value — use plt.scatter(x, y, c=z) where z is a third array
import matplotlib.pyplot as plt
import numpy as np
x = np.arange(30)
y = np.random.rand(30)
z = np.linspace(5,25,30)

plt.scatter(x, y, c=z, cmap='viridis')
plt.colorbar()
plt.title('color represents value')
plt.show()
```

```python
#Program-7
#Normalize a 10x10 random matrix to range 0–1 and verify using np.min() and np.max()
import numpy as np
import matplotlib.pyplot as plt
g = np.random.randint(0,100,(10,10))
g_normal = (g-g.min())/(g.max()-g.min())
plt.imshow(g)
plt.colorbar()
plt.plot()
print(np.min(g_normal),np.max(g_normal))
```
